```python
import shutil
from pathlib import Path
```

```python
# Paths
BASE_DIR  = Path(r"c:\Users\Usuario\Documents\Monroe County ORRI")
DEST_DIR  = BASE_DIR / "Excel Files"

SOURCES = [
    BASE_DIR / "_HG ORRI (1)",
    BASE_DIR / "_HG ORRI (2.0)",
    BASE_DIR / "UTIC files",
]

print(f"Destination: {DEST_DIR}")
```

```python
# Create flat destination folder (fresh each run)
if DEST_DIR.exists():
    print(f"Removing existing: {DEST_DIR}")
    shutil.rmtree(DEST_DIR)
DEST_DIR.mkdir()

seen_names = {}   # to handle duplicates

for source in SOURCES:
    for excel_path in sorted(source.rglob("*.xls*")):
        # Skip temp files created by Excel
        if excel_path.name.startswith("~$"):
            continue

        # Build prefix from path relative to BASE_DIR, replacing separators with " - "
        relative   = excel_path.relative_to(BASE_DIR)
        parts      = list(relative.parts[:-1])          # all folder parts, no filename
        prefix     = " - ".join(parts)                  # e.g. "UTIC files - UTIC00222-000 - Lease Acquisition Files"
        new_stem   = f"{prefix} - {excel_path.stem}"
        new_name   = f"{new_stem}{excel_path.suffix}"

        # Handle duplicates
        if new_name in seen_names:
            seen_names[new_name] += 1
            new_name = f"{new_stem}_{seen_names[new_name]}{excel_path.suffix}"
        else:
            seen_names[new_name] = 1

        dest_path = DEST_DIR / new_name
        shutil.copy2(excel_path, dest_path)
        print(f"  [OK]  {new_name}")

print(f"\nDone. {sum(seen_names.values())} Excel files collected in '{DEST_DIR.name}'.")
```
