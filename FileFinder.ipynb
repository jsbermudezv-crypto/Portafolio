```python
from pathlib import Path
import pandas as pd
```

```python
# Search keyword - edit this to change what you're looking for
SEARCH_KEY = "393"

# Root folder to search in
SEARCH_ROOT = Path(r"c:\Users\Usuario\Documents\Monroe County ORRI")
```

```python
# Search all files recursively whose name contains the keyword
matches = [
    str(p) for p in SEARCH_ROOT.rglob("*")
    if p.is_file() and SEARCH_KEY.lower() in p.name.lower()
]

print(f"Found {len(matches)} file(s) matching '{SEARCH_KEY}':")
for path in matches:
    print(path)
```

```python
# Show results as a DataFrame
results_df = pd.DataFrame({"file_path": matches})
results_df["file_name"] = results_df["file_path"].apply(lambda p: Path(p).name)
results_df["folder"]    = results_df["file_path"].apply(lambda p: str(Path(p).parent))
results_df[["file_name", "folder", "file_path"]]

print(results_df["file_path"][0])
```
