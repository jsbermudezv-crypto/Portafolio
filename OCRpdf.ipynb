```python
import pdfplumber
import pytesseract
import pandas as pd
import io
from IPython.display import display

import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
```

```python
# Paths
from pathlib import Path

BASE_DIR    = Path(r"c:\Users\Usuario\Documents\Monroe County ORRI\OtherFIles")
SOURCE_DIR  = BASE_DIR 
DEST_DIR    = BASE_DIR / "Andrea"

print(f"Source : {SOURCE_DIR}")
print(f"Dest   : {DEST_DIR}")
```

```python
# Batch OCR – mirrors the folder structure of others/ into others OCR/
# Output filename: original stem + " OCR.pdf"

from pypdf import PdfWriter, PdfReader

pdf_files = sorted(SOURCE_DIR.rglob("*.pdf"))
print(f"Found {len(pdf_files)} PDF(s) to process.\n")

for pdf_path in pdf_files:
    # Build mirrored destination path
    relative   = pdf_path.relative_to(SOURCE_DIR)
    dest_path  = DEST_DIR / relative.parent / (pdf_path.stem + " OCR.pdf")

    # Skip if already done
    if dest_path.exists():
        print(f"  [SKIP] {relative}  (already exists)")
        continue

    # Create destination subfolder if needed
    dest_path.parent.mkdir(parents=True, exist_ok=True)

    print(f"  [OCR]  {relative}")
    output_pdf = PdfWriter()

    try:
        with pdfplumber.open(pdf_path) as pdf:
            for page_num, page in enumerate(pdf.pages, start=1):
                img            = page.to_image(resolution=300).original
                ocr_pdf_bytes  = pytesseract.image_to_pdf_or_hocr(img, extension="pdf")
                reader         = PdfReader(io.BytesIO(ocr_pdf_bytes))
                output_pdf.add_page(reader.pages[0])

        with open(dest_path, "wb") as f:
            output_pdf.write(f)

        print(f"         -> saved: {dest_path.name}")

    except Exception as e:
        print(f"         [ERROR] {e}")

print("\nDone.")
```
